```python
import requests

url = "https://raw.githubusercontent.com/nlpaueb/multi-eurlex/master/data/eurovoc_descriptors.json"
eurovoc_concepts = requests.get(url).json()

for i, idx in enumerate(doc['labels'], 1):
    eurovoc_id = classlabel.int2str(idx)
    en_label = eurovoc_concepts[eurovoc_id]['en']
    fr_label = eurovoc_concepts[eurovoc_id]['fr']
    de_label = eurovoc_concepts[eurovoc_id]['de']
    print(f"{i}. Index {idx} → EuroVoc {eurovoc_id}")
    print(f"   EN: {en_label}")
    print(f"   FR: {fr_label}")
    print(f"   DE: {de_label}")
```

```python
# Load full test set and analyze
print("\n9. FULL TEST SET ANALYSIS")
print("-"*80)

dataset_full = load_dataset('coastalcph/multi_eurlex', 'all_languages',
                            split='test', label_level='all_levels',
                            trust_remote_code=True)

print(f"Total documents: {len(dataset_full)}")

# Check which languages are consistently available
lang_counts = {}
for doc in dataset_full[:100]:  # Sample first 100
    for lang, text in doc['text'].items():
        if text is not None:
            lang_counts[lang] = lang_counts.get(lang, 0) + 1

print("\nLanguage availability (first 100 docs):")
for lang, count in sorted(lang_counts.items(), key=lambda x: x[1], reverse=True):
    print(f"  {lang}: {count}/100 ({100*count/100:.0f}%)")
```
